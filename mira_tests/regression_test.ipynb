```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.arange(0, 10).reshape(-1, 1)
y = 2 * X.ravel() + np.random.default_rng(42).normal(0, 2, 10)

model = LinearRegression()
model.fit(X, y)

plt.scatter(X, y, label="Daten")
plt.plot(X, model.predict(X), label="Vorhersage")
plt.legend()
plt.title("Regression – Linear Regression")
plt.show()
```
